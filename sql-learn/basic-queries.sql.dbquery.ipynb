```sql
drop table Departments;
CREATE TABLE Departments (
    DeptID INT PRIMARY KEY,
    DeptName VARCHAR(50)
);
-- testing sql queries

CREATE TABLE Employees (
    EmpID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    ManagerID INT,
    Salary DECIMAL(10, 2),
    JoiningDate DATE,
    DeptID INT,
    FOREIGN KEY (DeptID) REFERENCES Departments(DeptID)
);

--inserting sample data into both tables
INSERT INTO Departments VALUES (1, 'IT'), (2, 'HR'), (3, 'Sales');
INSERT INTO Employees VALUES 
(101, 'John', 'Doe', NULL, 90000, '2022-01-15', 1),
(102, 'Jane', 'Smith', 101, 85000, '2022-02-10', 1),
(103, 'James', 'Brown', 101, 70000, '2022-03-05', 2),
(104, 'Emily', 'Davis', 102, 85000, '2022-01-20', 1),
(105, 'Michael', 'Wilson', 103, 95000, '2023-01-10', 3);


--full retrieval
SELECT * FROM Employees;

---specific retrieval
SELECT FirstName, LastName, Salary FROM Employees;


---using alias
SELECT FirstName, (Salary * 0.10) AS AnnualBonus FROM Employees;  --annual bonus is alias here
```
